```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
RAW_DATA_PATH = "data"

sales2017 = pd.read_csv(os.path.join(RAW_DATA_PATH, "Sales 2017.csv"))
sales2018 = pd.read_csv(os.path.join(RAW_DATA_PATH, "Sales 2018.csv"))
customers = pd.read_csv(os.path.join(RAW_DATA_PATH, "Customers.csv"))
products  = pd.read_csv(os.path.join(RAW_DATA_PATH, "Products.csv"))
regions   = pd.read_csv(os.path.join(RAW_DATA_PATH, "Region.csv"))
returns   = pd.read_csv(os.path.join(RAW_DATA_PATH, "Returns.csv"))
stores    = pd.read_csv(os.path.join(RAW_DATA_PATH, "Stores.csv"))
```

```python
tables = {
    "sales2017" : sales2017,
    "sales2018" : sales2018,
    "customers" : customers,
    "products" : products,
    "regions" : regions,
    "returns" : returns,
    "stores" : stores
}
```

```python
for key, df in tables.items():
    print(f"\t\t**{key}**")
    print(pd.DataFrame(df).head())
```

```python
for key, df in tables.items():
    print(f"\t**{key}**:\n{df.columns.tolist()}")
```

```python
for key, df in tables.items():
    print(f"{key}: {df.shape}")
```

```python
sales2017Copy = sales2017.copy().drop_duplicates()
sales2018Copy = sales2018.copy().drop_duplicates()
customersCopy = customers.copy().drop_duplicates()
productsCopy = products.copy().drop_duplicates()
regionsCopy = regions.copy().drop_duplicates()
returnsCopy = returns.copy().drop_duplicates()
storesCopy = stores.copy().drop_duplicates()
```

```python
tablesCopy = {
    "sales2017" : sales2017Copy,
    "sales2018" : sales2018Copy,
    "customers" : customersCopy,
    "products" : productsCopy,
    "regions" : regionsCopy,
    "returns" : returnsCopy,
    "stores" : storesCopy
}
```

```python
for key, df in tablesCopy.items():
    print(f"{key}: {df.shape}")
```

```python
for key, df in tablesCopy.items():
    print(f"\t\t== {key}: {df.shape} ==")
    print(f"\n{df.info()}")
```

```python
sales2017Copy[["transaction_date", "stock_date"]] = sales2017Copy[["transaction_date", "stock_date"]].apply(pd.to_datetime)
sales2017Copy.info()
```

```python
sales2018Copy[["transaction_date", "stock_date"]] = sales2018Copy[["transaction_date", "stock_date"]].apply(pd.to_datetime)
sales2018Copy.info()
```

```python
customersCopy[["birthdate", "acct_open_date"]] = customersCopy[["birthdate", "acct_open_date"]].apply(pd.to_datetime)
customersCopy.info()
```

```python
returnsCopy["return_date"] = pd.to_datetime(returnsCopy["return_date"])
returnsCopy.info()
```

```python
storesCopy[["first_opened_date", "last_remodel_date"]] = storesCopy[["first_opened_date", "last_remodel_date"]].apply(pd.to_datetime)
storesCopy.info()
```

```python
Sales = pd.concat([sales2017Copy, sales2018Copy]).drop_duplicates()
```
